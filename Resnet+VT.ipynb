```python
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Concatenate, Input
from vit_keras import vit, utils
from tensorflow.keras.models import Model
```

```python
# Define input layer
input_layer = Input(shape=(224, 224, 3))
```

```python
# Initialize and fine-tune ResNet model
resnet = ResNet50(include_top=False, weights='imagenet')
for layer in resnet.layers[:-5]:
    layer.trainable = False
resnet_features = resnet(input_layer)
resnet_features = GlobalAveragePooling2D()(resnet_features)
```

```python
# Initialize Vision Transformer model
vit_model = vit.vit_b32(image_size=224, pretrained=True, include_top=True)
vit_features = vit_model(input_layer, training = True)
vit_features = GlobalAveragePooling2D()(vit_features)
```
